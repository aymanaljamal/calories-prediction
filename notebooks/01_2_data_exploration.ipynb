## 7. Correlation Matrix
## مصفوفة الارتباط

Correlation measures the strength and direction of a linear relationship between numerical variables.

يقيس الارتباط قوة واتجاه العلاقة الخطية بين المتغيرات الرقمية.

Values close to +1 indicate a strong positive linear relationship, while values close to -1 indicate a strong negative relationship. Values close to 0 indicate a weak linear relationship.

القيم القريبة من +1 تشير إلى علاقة خطية موجبة قوية، والقيم القريبة من -1 تشير إلى علاقة سالبة قوية، بينما القيم القريبة من 0 تشير إلى علاقة خطية ضعيفة.

```python
correlation = get_correlation(df)

display(correlation)
```

## 8. Correlation with Target
## الارتباط مع المتغير المستهدف

Here we calculate how each numerical feature is linearly correlated with `Calories Burned`.

هنا نحسب مدى ارتباط كل متغير رقمي خطيًا مع `Calories Burned`.

```python
target_column = "Calories Burned"

target_correlations = (
    correlation[target_column]
    .drop(target_column)
    .sort_values(ascending=False)
)

display(
    target_correlations.to_frame(name="Correlation")
)
```

### Interpretation | التفسير

Correlation only measures linear relationships. A low correlation does not necessarily mean that a feature has no predictive value because the relationship may be nonlinear or involve interactions with other variables.

الارتباط يقيس العلاقات الخطية فقط. لذلك فإن الارتباط المنخفض لا يعني بالضرورة أن المتغير غير مفيد للتنبؤ، فقد تكون العلاقة غير خطية أو تعتمد على تفاعل المتغير مع متغيرات أخرى.

## 9. Absolute Target Correlations
## قوة الارتباط مع المتغير المستهدف

We sort features by the absolute value of their correlation with the target.

نرتب المتغيرات حسب القيمة المطلقة للارتباط مع المتغير المستهدف، حتى نعرف قوة العلاقة بغض النظر عن اتجاهها.

```python
absolute_target_correlations = (
    target_correlations.abs()
    .sort_values(ascending=False)
)

display(
    absolute_target_correlations.to_frame(
        name="Absolute Correlation"
    )
)
```

## 10. Target Variable Statistics
## إحصاءات المتغير المستهدف

The target variable is `Calories Burned`. We examine its distribution numerically before building a regression model.

المتغير المستهدف هو `Calories Burned`. نقوم بدراسة توزيعه رقميًا قبل بناء نموذج الانحدار.

```python
target_analysis = analyze_target(
    df,
    target_column
)

display(
    target_analysis.to_frame(name="Value")
)
```

## 11. Target Distribution
## توزيع المتغير المستهدف

We inspect the minimum, maximum, mean, and median of calories burned.

نستعرض الحد الأدنى والأقصى والمتوسط والوسيط للسعرات المحروقة.

```python
target = df[target_column]

print(f"Minimum : {target.min():.2f}")
print(f"Maximum : {target.max():.2f}")
print(f"Mean    : {target.mean():.2f}")
print(f"Median  : {target.median():.2f}")
print(f"Std     : {target.std():.2f}")
```

## 12. Target Quantiles
## النسب المئوية للمتغير المستهدف

Quantiles help us understand how the target values are distributed across the dataset.

تساعدنا النسب المئوية على فهم كيفية توزيع قيم المتغير المستهدف داخل البيانات.

```python
quantiles = target.quantile([
    0.01,
    0.05,
    0.10,
    0.25,
    0.50,
    0.75,
    0.90,
    0.95,
    0.99
])

quantiles.index = [
    "1%",
    "5%",
    "10%",
    "25%",
    "50%",
    "75%",
    "90%",
    "95%",
    "99%"
]

display(quantiles.to_frame(name="Calories Burned"))
```

## 13. Gender Analysis
## تحليل الجنس

We compare the number of observations and calorie statistics between gender categories.

نقارن عدد السجلات وإحصاءات السعرات الحرارية بين فئات الجنس.

```python
if "Gender" in df.columns:
    gender_summary = (
        df.groupby("Gender")[target_column]
        .agg([
            "count",
            "mean",
            "median",
            "min",
            "max",
            "std"
        ])
        .round(2)
    )

    display(gender_summary)
else:
    print("Gender column not found.")
```

## 14. Running Activity Summary
## ملخص نشاط الجري

The following variables describe the running activity:

المتغيرات التالية تصف نشاط الجري:

- Running Time(min) — مدة الجري بالدقائق
- Running Speed(km/h) — سرعة الجري بالكيلومتر/ساعة
- Distance(km) — المسافة بالكيلومتر
- Average Heart Rate — متوسط معدل ضربات القلب

```python
activity_columns = [
    "Running Time(min)",
    "Running Speed(km/h)",
    "Distance(km)",
    "Average Heart Rate"
]

activity_columns = [
    column for column in activity_columns
    if column in df.columns
]

display(df[activity_columns].describe().T)
```

## 15. Body Measurements Summary
## ملخص قياسات الجسم

We also examine age, height, weight, and BMI.

نقوم أيضًا بدراسة العمر والطول والوزن ومؤشر كتلة الجسم.

```python
body_columns = [
    "Age",
    "Height(cm)",
    "Weight(kg)",
    "BMI"
]

body_columns = [
    column for column in body_columns
    if column in df.columns
]

display(df[body_columns].describe().T)
```

## 16. Calories by Running Time
## السعرات مقابل مدة الجري

We inspect the average calories burned for different running-time groups.

نستعرض متوسط السعرات المحروقة حسب مجموعات مدة الجري.

```python
if "Running Time(min)" in df.columns:
    time_groups = pd.qcut(
        df["Running Time(min)"],
        q=4,
        duplicates="drop"
    )

    time_analysis = (
        df.groupby(time_groups, observed=True)[target_column]
        .agg(["count", "mean", "median", "min", "max"])
        .round(2)
    )

    display(time_analysis)
else:
    print("Running Time column not found.")
```

## 17. Calories by Distance
## السعرات مقابل المسافة

We divide the distance values into four groups and compare their calorie statistics.

نقسم قيم المسافة إلى أربع مجموعات ونقارن إحصاءات السعرات بينها.

```python
if "Distance(km)" in df.columns:
    distance_groups = pd.qcut(
        df["Distance(km)"],
        q=4,
        duplicates="drop"
    )

    distance_analysis = (
        df.groupby(distance_groups, observed=True)[target_column]
        .agg(["count", "mean", "median", "min", "max"])
        .round(2)
    )

    display(distance_analysis)
else:
    print("Distance column not found.")
```

## 18. Calories by Heart Rate
## السعرات مقابل معدل ضربات القلب

We group average heart rate into quartiles and compare calorie statistics.

نقسم متوسط معدل ضربات القلب إلى أربع مجموعات ونقارن إحصاءات السعرات الحرارية.

```python
if "Average Heart Rate" in df.columns:
    heart_rate_groups = pd.qcut(
        df["Average Heart Rate"],
        q=4,
        duplicates="drop"
    )

    heart_rate_analysis = (
        df.groupby(heart_rate_groups, observed=True)[target_column]
        .agg(["count", "mean", "median", "min", "max"])
        .round(2)
    )

    display(heart_rate_analysis)
else:
    print("Average Heart Rate column not found.")
```

## 19. Calories by Weight
## السعرات مقابل الوزن

We group weight values into quartiles and compare the target variable.

نقسم قيم الوزن إلى أربع مجموعات ونقارن المتغير المستهدف بينها.

```python
if "Weight(kg)" in df.columns:
    weight_groups = pd.qcut(
        df["Weight(kg)"],
        q=4,
        duplicates="drop"
    )

    weight_analysis = (
        df.groupby(weight_groups, observed=True)[target_column]
        .agg(["count", "mean", "median", "min", "max"])
        .round(2)
    )

    display(weight_analysis)
else:
    print("Weight column not found.")
```
